# Python Operators

Operators are symbols or keywords used to perform operations on values and variables.

## 1. Arithmetic Operators

Used for mathematical operations.

| Operator | Meaning        | Example  |
| -------- | -------------- | -------- |
| `+`      | Addition       | `5 + 2`  |
| `-`      | Subtraction    | `5 - 2`  |
| `*`      | Multiplication | `5 * 2`  |
| `/`      | Division       | `5 / 2`  |
| `%`      | Remainder      | `5 % 2`  |
| `**`     | Power          | `5 ** 2` |
| `//`     | Floor Division | `5 // 2` |

---

## 2. Assignment Operators

Used to assign or update values.

| Operator | Example   |
| -------- | --------- |
| `=`      | `x = 5`   |
| `+=`     | `x += 2`  |
| `-=`     | `x -= 2`  |
| `*=`     | `x *= 2`  |
| `/=`     | `x /= 2`  |
| `%=`     | `x %= 2`  |
| `**=`    | `x **= 2` |
| `//=`    | `x //= 2` |

`x += 2` means `x = x + 2`.

---

## 3. Comparison Operators

Used to compare two values. Result is `True` or `False`.

| Operator | Meaning               |
| -------- | --------------------- |
| `==`     | Equal                 |
| `!=`     | Not equal             |
| `>`      | Greater than          |
| `<`      | Less than             |
| `>=`     | Greater than or equal |
| `<=`     | Less than or equal    |

---

## 4. Logical Operators

Used to combine conditions.

| Operator | Meaning                             |
| -------- | ----------------------------------- |
| `and`    | Both conditions must be True        |
| `or`     | At least one condition must be True |
| `not`    | Reverses the result                 |

Example:

```python
age = 20
print(age > 18 and age < 25)
```

---

## 5. Identity Operators

Used to check whether two variables refer to the **same object**.

| Operator | Meaning           |
| -------- | ----------------- |
| `is`     | Same object       |
| `is not` | Different objects |

Example:

```python
x = [1, 2]
y = x

print(x is y)  # True
```

**Note:** `==` checks values, while `is` checks object identity.

---

## 6. Membership Operators

Used to check whether a value exists in a sequence.

| Operator | Meaning              |
| -------- | -------------------- |
| `in`     | Value exists         |
| `not in` | Value does not exist |

Example:

```python
numbers = [10, 20, 30]

print(20 in numbers)      # True
print(50 not in numbers)  # True
```

---

## 7. Bitwise Operators

Used to perform operations on the **bits** of integers.

| Operator | Meaning     |    |
| -------- | ----------- | -- |
| `&`      | AND         |    |
| `        | `           | OR |
| `^`      | XOR         |    |
| `~`      | NOT         |    |
| `<<`     | Left shift  |    |
| `>>`     | Right shift |    |

Example:

```python
print(5 & 3)  # 1
print(5 | 3)  # 7
print(5 ^ 3)  # 6
```

### Quick Revision

* Arithmetic → Mathematical operations
* Assignment → Assign values
* Comparison → Compare values
* Logical → Combine conditions
* Identity → Compare objects
* Membership → Check existence
* Bitwise → Work with bits
