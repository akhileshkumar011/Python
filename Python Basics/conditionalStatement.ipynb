#(Grade Calculator
Take marks as input.
90 or above → A
75–89 → B
50–74 → C
40–49 → D
Below 40 → Fail)

```python
marks = int(input("enter the marks"))
if marks >= 90 :
    print("A")
elif marks >=75 and marks <=89 :
    print("B")  
elif marks >=50 and marks <=74 :
    print("C")    
elif marks >=40 and marks <=49 :
    print("D")      
elif marks <49  :
    print("Fail") 
```
