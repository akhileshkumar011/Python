```python
## list 

lst = ["apple","banana","grapes"]
print(lst)

## access the element of the list
print(lst[0])

## list allow the duplicates 
lsit = ["apple","banana","grapes","apple","banana"]
print(lsit[4])
```

```python
## list length
lst = ["apple","banana","grapes"]
print(len(lst))
```

```python
## list items has any data type
lst = ["ram","shyama","aman"]
lst1 = [1,2,3,4,5]
lst2 = [True,False,True]

##  a list can have different data types
lst3 = ["akhil",1,True,5.6] 
print(len(lst3))
```

```python
## type of the list items 

lst = ["apple","banana","grapes"]
lst1 = ["patel",1,True,5.6]
print(type(lst))
print(type(lst1))
```

```python
## list() constructor 
thlist = list(("apple","banana","grapes")) ### note the double round-brackets
print(thlist)
```

```python
## access the list items 
lst = ["apple","banana","grapes"]
print(lst[0])

## negative indexing 
## means start from the end 
print(lst[-1])



```

```python
## list range 

lst = ["apple","banana","grapes"]
print(lst[1:2])


lst1 = list(("apple","banana","grapes","mango","kiwi"))
print(lst1[:5])
print(lst1[2:])


##Range of Negative Indexes
print(lst1[-4:-1])
```

```python
##Check if Item Exists
lst1 = list(("apple","banana","grapes","mango","kiwi"))
print("banana" in lst1)
```

```python
##insert items

thatlist = ['her','his','whyyy']
thatlist.insert(2,"people")
print(thatlist)
```

```python
## Add list items

##Append()
##To add an item to the end of the list, use the append() method

thatlist = ["apple","cherry","banana"]
thatlist.append("orange")
print(thatlist)


```

```python
##Insert Items
##insert()
##to insert item at specific position we use insert() function

thatlist = ["apple","cherry","banana"]
thatlist.insert(2,'papaya')
print(thatlist)
```

```python
##Extend List
##To append elements from another list to the current list, use the extend() method.
thislist = ['board','switch','button']
thatlist = ['fan','AC','White Board']
thislist.extend(thatlist)
print(thislist)


##he extend() method does not have to append lists, you can add any iterable object (tuples, sets, dictionaries etc.).

theselist = ('light','desk')
thislist.extend(theselist)
print(thislist)
```
