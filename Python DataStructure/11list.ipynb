```python
##Extend List
##To append elements from another list to the current list, use the extend() method.
thislist = ['board','switch','button']
thatlist = ['fan','AC','White Board']
thislist.extend(thatlist)
print(thislist)


##he extend() method does not have to append lists, you can add any iterable object (tuples, sets, dictionaries etc.).

theselist = ('light','desk')
thislist.extend(theselist)
print(thislist)
```

# Remove List Items
## Remove()
- The remove() method removes the specified item.



```python

thislist = ["apple","banana","tharabhaijogindar",'rajesh',"ferrari"]
thislist.remove('apple')
print(thislist)


```

## pop()
- The pop() method removes the specified index.
- if you do not specify the index, the pop() method removes the last item.

```python

## Pop

thislist = ["apple","banana","tharabhaijogindar",'rajesh',"ferrari"]
thislist.pop(1)
print(thislist)
thislist.pop()
print(thislist)
```

## del()
- The del keyword also removes the specified index:
- The del keyword can also delete the list completely.

```python

thislist = ["apple","banana","tharabhaijogindar",'rajesh',"ferrari"]
del thislist[1]
print(thislist)

del thislist
```

## Clear the list
- the clear() method is empties the list
- the list will still remain but there is no content in it 


```python
thislist = ["apple","banana","tharabhaijogindar",'rajesh',"ferrari"]
thislist.clear()
print(thislist)
```

# Loop Lists 
## Loop Through a List
 - You can loop through the list items by using a for loop:
## Loop Through the Index Numbers
 - You can also loop through the list items by referring to their index number.
 - Use the range() and len() functions to create a suitable iterable.

```python
thislist = ["apple","banana","tharabhaijogindar",'rajesh',"ferrari"]
for x in thislist :
 print(x)

for  i in range(len(thislist)) :
 print(thislist[i])
  

 
```

## Using a While Loop


```python
thislist = ["apple","banana","tharabhaijogindar",'rajesh',"ferrari"]

i = 0

while i < len(thislist) :
    print(thislist[i])
    i = i + 1
```
