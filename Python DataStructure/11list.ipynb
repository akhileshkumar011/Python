# Remove List Items
## Remove()
- The remove() method removes the specified item.



```python

thislist = ["apple","banana","tharabhaijogindar",'rajesh',"ferrari"]
thislist.remove('apple')
print(thislist)


```

## pop()
- The pop() method removes the specified index.
- if you do not specify the index, the pop() method removes the last item.

```python

## Pop

thislist = ["apple","banana","tharabhaijogindar",'rajesh',"ferrari"]
thislist.pop(1)
print(thislist)
thislist.pop()
print(thislist)
```

## del()
- The del keyword also removes the specified index:
- The del keyword can also delete the list completely.

```python

thislist = ["apple","banana","tharabhaijogindar",'rajesh',"ferrari"]
del thislist[1]
print(thislist)

del thislist
```

## Clear the list
- the clear() method is empties the list
- the list will still remain but there is no content in it 


```python
thislist = ["apple","banana","tharabhaijogindar",'rajesh',"ferrari"]
thislist.clear()
print(thislist)
```

```python
thislist = ["apple","banana","tharabhaijogindar",'rajesh',"ferrari"]
for x in thislist :
 print(x)

for  i in range(len(thislist)) :
 print(thislist[i])
  

 
```

```python
thislist = ["apple","banana","tharabhaijogindar",'rajesh',"ferrari"]

i = 0

while i < len(thislist) :
    print(thislist[i])
    i = i + 1
```

# Sort Lists
## Sort List Alphanumerically
 - List objects have a sort() method that will sort the list alphanumerically, ascending, by default:



```python
list = ["orange", "mango", "kiwi", "pineapple", "banana"]
list.sort()
print(list)
```

## Sort Descending
- To sort descending, use the keyword argument reverse = True

```python
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
thislist.sort(reverse = True)
print(thislist)
```

## Customize Sort Function
- You can also customize your own function by using the keyword argument key = function.
- The function will return a number that will be used to sort the list (the lowest number first):

```python
def func(n):
    return abs(n-50)
thislst = [100, 50, 65, 82, 23]
thislst.sort(key = func)
print(thislst)
```

## Case Insensitive Sort
- By default the sort() method is case sensitive, resulting in all capital letters being sorted before lower case letters:

```python
listfirst = ["apple", "Aman", "shyama", "Grapes"]
listfirst.sort()
print(listfirst)
```

## Reverse Order
- What if you want to reverse the order of a list, regardless of the alphabet?
- The reverse() method reverses the current sorting order of the elements.

```python
list = [10,20,40,50]
list.reverse()
print(list)
```

# Copy Lists
## Use the copy() method
- You can use the built-in List method copy() to copy a list.


```python
list = ["orange", "mango", "kiwi", "pineapple", "banana"]
list1 = list.copy()
print(list1)
```

## Use the list() method
- Another way to make a copy is to use the built-in method list().


```python
list2 = ["orange", "mango", "kiwi", "pineapple", "banana"]
list1 = list(list2)
print(list1)
```

## Use the slice Operator
- You can also make a copy of a list by using the : (slice) operator.

```python
list = ["orange", "mango", "kiwi", "pineapple", "banana"]
list1 = list[:]
print(list1)
```

# Python - Join Lists
## Join Two Lists
- There are several ways to join, or concatenate, two or more lists in Python.
- One of the easiest ways are by using the + operator.

## Append list2 into list1:

```python
lst1 = ["orange", "mango", "kiwi", "pineapple", "banana"]
lst2 = [50,60,70,80,90]
lst3 = lst1 + lst2
print(lst3)

```

```python
lst1 = ["orange", "mango", "kiwi", "pineapple", "banana"]
lst2 = [50,60,70,80,90]
for x in lst2:
    lst1.append(x)
print(lst1)   
```

## Extend()
-  you can use the extend() method, where the purpose is to add elements from one list to another list:

```python
lst1 = ["orange", "mango", "kiwi", "pineapple", "banana"]
lst2 = [50,60,70,80,90]
lst1.extend(lst2)
print(lst1)
```
